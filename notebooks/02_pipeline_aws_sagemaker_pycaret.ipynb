MODELO DE MACHINE LEARNING

Vamos a complementar tu pipeline con scikit-learn puro, usando el mejor modelo que salga de pycaret como base. 

```python
from pycaret.regression import setup, create_model, pull, get_config

# 1. Setup con TODAS las columnas (data cruda completa)
experiment = setup(data=df, target='ivc_n', session_id=123)

# 2. Crear directamente el modelo ganador (LightGBM)
best_model = create_model('lightgbm')
metricas_lightgbm = pull()
display(metricas_lightgbm)
metricas_lightgbm.to_csv('metricas_lightgbm_tesis.csv', index=True)

print(type(best_model))
print(best_model)

# 3. Obtener los datos ya transformados por pycaret (mismo preprocesamiento que usó el modelo)
X = get_config('X')
y = get_config('y')

print(X.shape, y.shape)
```

3. Entrenar con múltiples splits (80/20, 90/10, 70/30)

```python
from sklearn.model_selection import train_test_split
from sklearn.base import clone
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

splits = {
    '70/30': 0.30,
    '80/20': 0.20,
    '90/10': 0.10
}

resultados_splits = []

for nombre_split, test_size in splits.items():
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=123
    )
    
    modelo = clone(best_model)  # copia limpia del mejor modelo, sin entrenar
    modelo.fit(X_train, y_train)
    
    y_pred = modelo.predict(X_test)
    
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    
    resultados_splits.append({
        'split': nombre_split,
        'n_train': len(X_train),
        'n_test': len(X_test),
        'R2': r2,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse
    })

df_splits = pd.DataFrame(resultados_splits)
display(df_splits)
df_splits.to_csv('resultados_splits_tesis.csv', index=False)
```

4. Validación cruzada (K-Fold)

```python
from sklearn.model_selection import KFold, cross_validate

# Probar distintos números de folds (común: 5 y 10)
folds_a_probar = [5, 10]
resultados_kfold = []

for k in folds_a_probar:
    kf = KFold(n_splits=k, shuffle=True, random_state=123)
    
    modelo = clone(best_model)
    
    cv_results = cross_validate(
        modelo, X, y,
        cv=kf,
        scoring=['r2', 'neg_mean_absolute_error', 'neg_mean_squared_error'],
        return_train_score=False
    )
    
    resultados_kfold.append({
        'k_folds': k,
        'R2_mean': cv_results['test_r2'].mean(),
        'R2_std': cv_results['test_r2'].std(),
        'MAE_mean': -cv_results['test_neg_mean_absolute_error'].mean(),
        'MSE_mean': -cv_results['test_neg_mean_squared_error'].mean(),
        'RMSE_mean': np.sqrt(-cv_results['test_neg_mean_squared_error']).mean()
    })

df_kfold = pd.DataFrame(resultados_kfold)
display(df_kfold)
df_kfold.to_csv('resultados_kfold_tesis.csv', index=False)
```

5. Pipeline completo de scikit-learn (encapsulando todo)

Para que quede como un pipeline formal reutilizable (ideal para el despliegue posterior):

```python
from sklearn.pipeline import Pipeline

# Si tu best_model ya es un pipeline de pycaret, esto lo mantiene encapsulado
pipeline_final = Pipeline([
    ('regressor', clone(best_model))
])

# Entrenar con el split que prefieras usar como definitivo (ej. 80/20)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=123
)

pipeline_final.fit(X_train, y_train)

y_pred = pipeline_final.predict(X_test)
print(f"R2 final: {r2_score(y_test, y_pred):.4f}")
print(f"MAE final: {mean_absolute_error(y_test, y_pred):.4f}")
print(f"RMSE final: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```
