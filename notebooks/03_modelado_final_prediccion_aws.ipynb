Comparar modelo acorde train y test

```python
from pycaret.regression import get_config
from sklearn.model_selection import train_test_split
from sklearn.base import clone
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

X = get_config('X')
y = get_config('y')
print("Shape X, y:", X.shape, y.shape)  # debería confirmar (43408, 76) y (43408,)

splits = {'70/30': 0.30, '80/20': 0.20, '90/10': 0.10, '60/40': 0.40}
resultados_splits = []
modelos_por_split = {}

for nombre_split, test_size in splits.items():
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=123
    )
    modelo = clone(modelo_final)
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)

    resultados_splits.append({
        'split': nombre_split, 'n_train': len(X_train), 'n_test': len(X_test),
        'R2': r2_score(y_test, y_pred),
        'MAE': mean_absolute_error(y_test, y_pred),
        'MSE': mean_squared_error(y_test, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_test, y_pred))
    })
    modelos_por_split[nombre_split] = modelo

df_splits = pd.DataFrame(resultados_splits)
display(df_splits)
df_splits.to_csv('comparacion_splits_hiperparametros_finales_tesis.csv', index=False)

mejor_split = df_splits.loc[df_splits['RMSE'].idxmin(), 'split']
print(f"Mejor partición según RMSE: {mejor_split}")
```

El proceso de ajuste de hiperparámetros (fine-tuning) aplicado al modelo LightGBM produjo una mejora consistente y estadísticamente relevante respecto al modelo base. Al evaluar el modelo optimizado (max_depth=15, num_leaves=20, n_estimators=200) bajo cuatro esquemas de partición train/test (60/40, 70/30, 80/20 y 90/10), se observó un desempeño estable con coeficientes de determinación (R²) en el rango de 0.891 a 0.900, frente al 0.8856 obtenido por el modelo con hiperparámetros por defecto — una mejora de entre 0.5 y 1.4 puntos porcentuales, acompañada de una reducción sistemática del error (RMSE entre 0.0247 y 0.0255, frente a 0.0260 del modelo base).

La baja dispersión de resultados entre las distintas proporciones de partición (variación de apenas 0.9 puntos de R² entre el esquema más exigente —60/40— y el más favorable —90/10—) constituye evidencia de que el modelo no depende de una configuración particular de datos para generalizar, sino que captura relaciones estructurales robustas entre las variables predictoras (socioeconómicas, catastrales y climáticas) y el Índice de Vulnerabilidad Climática. Este comportamiento es consistente con los resultados obtenidos en la validación cruzada k-fold reportada previamente, reforzando la fiabilidad del modelo antes de proceder a su finalización y despliegue.

En términos prácticos, estos resultados validan la decisión metodológica de restringir la complejidad del modelo (menor num_leaves respecto al valor por defecto), evidenciando que una configuración más conservadora no solo mejora la precisión predictiva, sino que previsiblemente favorece la capacidad de generalización del modelo — un atributo especialmente relevante de cara al ejercicio de transferencia espacial hacia Medellín y Cali, donde el modelo deberá operar sobre datos con distribuciones distintas a las observadas durante el entrenamiento en Bogotá.

```python
import geopandas as gpd
from shapely.geometry import Point

# Crear geometría de puntos a partir de LATITUD/LONGITUD
geometry_puntos = [Point(xy) for xy in zip(predicciones['LONGITUD'], predicciones['LATITUD'])]

gdf_predicciones = gpd.GeoDataFrame(
    predicciones,
    geometry=geometry_puntos,
    crs='EPSG:4326'  # WGS84, sistema estándar de lat/long
)

print(gdf_predicciones.crs)
print(gdf_manzanas.crs)
```

Paso 1 — Alinear el sistema de coordenadas

```python
gdf_predicciones = gdf_predicciones.to_crs(gdf_manzanas.crs)

print("CRS predicciones:", gdf_predicciones.crs)
print("CRS shapefile:", gdf_manzanas.crs)
print("¿Coinciden?", gdf_predicciones.crs == gdf_manzanas.crs)
```

Paso 2 — Cruce espacial (point-in-polygon)

```python
gdf_final = gpd.sjoin(
    gdf_predicciones,
    gdf_manzanas[['MANCODIGO', 'SECCODIGO', 'geometry']],
    how='left',
    predicate='within'
)

print("Shape resultado:", gdf_final.shape)
print("Sin manzana asignada:", gdf_final['MANCODIGO'].isna().sum())
print("Con manzana asignada:", gdf_final['MANCODIGO'].notna().sum())
display(gdf_final[['MANCODIGO', 'SECCODIGO', 'LATITUD', 'LONGITUD', 'IVC_predicho']].head(10))
```

Paso 3 — Validar el cruce (obligatorio antes de exportar)

```python
# Si hay puntos sin asignar, prueba con 'intersects' (más tolerante en bordes)
if gdf_final['MANCODIGO'].isna().sum() > 0:
    gdf_final_v2 = gpd.sjoin(
        gdf_predicciones, gdf_manzanas[['MANCODIGO', 'SECCODIGO', 'geometry']],
        how='left', predicate='intersects'
    )
    print("Sin asignar (intersects):", gdf_final_v2['MANCODIGO'].isna().sum())
    if gdf_final_v2['MANCODIGO'].isna().sum() < gdf_final['MANCODIGO'].isna().sum():
        gdf_final = gdf_final_v2
        print("Se usó 'intersects' por mejor cobertura")
```

Paso 4 — Revisar duplicados (un punto puede caer en 2 polígonos si comparten borde)

```python
duplicados = gdf_final.index.duplicated().sum()
print("Filas duplicadas por el join:", duplicados)

if duplicados > 0:
    gdf_final = gdf_final[~gdf_final.index.duplicated(keep='first')]
    print("Duplicados eliminados. Shape final:", gdf_final.shape)
```

Paso 5 — Limpiar columnas antes de exportar

```python
gdf_final_export = gdf_final.drop(columns=['index_right'], errors='ignore')

# Verifica que las columnas clave estén presentes
print(gdf_final_export[['MANCODIGO', 'SECCODIGO', 'IVC_predicho']].isna().sum())
```

```python
gdf_final_export.to_file('manzanas_ivc_predicho.shp')
print("Shapefile guardado localmente")
```

```python
import os
import zipfile
import glob
import boto3
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import awswrangler as wr
from pycaret.regression import load_model, predict_model
import matplotlib.pyplot as plt
import contextily as ctx
from matplotlib_scalebar.scalebar import ScaleBar

# --- 1. CONFIGURACIÓN INICIAL ---
bucket_entrada = 'mapastiposhapemanzanascalimedellin'
bucket_salida = 'prediccionesbogota'

# --- 2. CARGA DEL MODELO Y DATOS TABULARES ---
print("Descargando modelo entrenado...")
wr.s3.download(path='s3://databogota/modelos/modelo_ivc_bogota_final_tuned.pkl', local_file='modelo_ivc_bogota_final_tuned.pkl')
modelo_produccion = load_model('modelo_ivc_bogota_final_tuned')

print("Cargando datos CSV de Medellín y Cali...")
df_medellin = wr.s3.read_csv(f's3://{bucket_entrada}/CLEAN DATA PARA MODELAR/MEDELLIN CALI/df_medellin_ivc_filtrado.csv')
df_cali = wr.s3.read_csv(f's3://{bucket_entrada}/CLEAN DATA PARA MODELAR/MEDELLIN CALI/df_cali_ivc_filtrado.csv')

# --- 3. ALINEACIÓN INTELIGENTE DE TIPOS DE DATOS ---
def alinear_con_modelo(df_nuevo, modelo):
    df_resultado = df_nuevo.copy()
    try:
        columnas_esperadas = modelo.steps[0][1].feature_names_in_
    except Exception:
        pipeline_obj = modelo.named_steps.get('actual_estimator', modelo)
        columnas_esperadas = getattr(pipeline_obj, 'feature_names_in_', df_nuevo.columns)

    for col in columnas_esperadas:
        if col not in df_resultado.columns:
            if col in ['NMB_LC_CM', 'TP_LC_CM']:
                df_resultado[col] = 'Desconocido'
            else:
                df_resultado[col] = 0.0 
                
    return df_resultado[[c for c in columnas_esperadas if c in df_resultado.columns]]

df_medellin_listo = alinear_con_modelo(df_medellin, modelo_produccion)
df_cali_listo = alinear_con_modelo(df_cali, modelo_produccion)

for df_orig, df_dest in zip([df_medellin, df_cali], [df_medellin_listo, df_cali_listo]):
    if 'LATITUD' in df_orig.columns:
        df_dest['LATITUD'] = df_orig['LATITUD']
    if 'LONGITUD' in df_orig.columns:
        df_dest['LONGITUD'] = df_orig['LONGITUD']

# --- 4. INFERENCIA DEL MODELO ---
print("Ejecutando predicciones para Medellín...")
pred_medellin = predict_model(modelo_produccion, data=df_medellin_listo)
pred_medellin.rename(columns={'prediction_label': 'IVC_predicho'}, inplace=True)

print("Ejecutando predicciones para Cali...")
pred_cali = predict_model(modelo_produccion, data=df_cali_listo)
pred_cali.rename(columns={'prediction_label': 'IVC_predicho'}, inplace=True)

# --- 5. UBICACIÓN ESPACIAL (PUNTOS) ---
print("Mapeando coordenadas espaciales...")
gdf_pred_med = gpd.GeoDataFrame(
    pred_medellin, 
    geometry=[Point(xy) for xy in zip(pred_medellin['LONGITUD'], pred_medellin['LATITUD'])], 
    crs='EPSG:4326'
)
gdf_pred_cali = gpd.GeoDataFrame(
    pred_cali, 
    geometry=[Point(xy) for xy in zip(pred_cali['LONGITUD'], pred_cali['LATITUD'])], 
    crs='EPSG:4326'
)

# --- 6. EXTRACCIÓN DE SHAPEFILES Y GEODATABASES ---
print("Procesando geometrías base de Medellín...")
wr.s3.download(f's3://{bucket_entrada}/shp_lote_del_predio.zip', local_file='medellin.zip')
if os.path.exists('shp_medellin'):
    import shutil
    shutil.rmtree('shp_medellin')
os.makedirs('shp_medellin', exist_ok=True)
with zipfile.ZipFile('medellin.zip', 'r') as z:
    z.extractall('shp_medellin')

rutas_shp_med = glob.glob('shp_medellin/**/*.shp', recursive=True)
if rutas_shp_med:
    gdf_medellin_shp = gpd.read_file(rutas_shp_med[0])
    print(f"✅ Shapefile de Medellín cargado correctamente.")
else:
    raise FileNotFoundError("❌ No se encontró ningún archivo .shp en Medellín.")

print("Procesando geodatabase base de Cali (.rar)...")
wr.s3.download(f's3://{bucket_entrada}/GEODATABASE CATASTRAL CON CORTE A ENERO 2024.rar', local_file='cali.rar')
if os.path.exists('shp_cali'):
    import shutil
    shutil.rmtree('shp_cali')
os.makedirs('shp_cali', exist_ok=True)

# Descompresión mediante el binario unrar instalado en el entorno
os.system('unrar x -y cali.rar shp_cali/')

rutas_gdb = glob.glob('shp_cali/**/*.gdb', recursive=True)
rutas_shp = glob.glob('shp_cali/**/*.shp', recursive=True)

if rutas_gdb:
    gdf_cali_shp = gpd.read_file(rutas_gdb[0], driver='OpenFileGDB')
    print(f"✅ Geodatabase de Cali cargada exitosamente: {rutas_gdb[0]}")
elif rutas_shp:
    gdf_cali_shp = gpd.read_file(rutas_shp[0])
    print(f"✅ Shapefile de Cali cargado exitosamente: {rutas_shp[0]}")
else:
    raise FileNotFoundError("❌ No se encontró ningún archivo .gdb o .shp tras descomprimir el .rar de Cali.")

# Validación y asignación de CRS para Cali si viene indefinido
if gdf_cali_shp.crs is None:
    if gdf_cali_shp.total_bounds[2] > 180:
        gdf_cali_shp.set_crs('EPSG:3115', inplace=True) # Magna-Sirgas / Colombia West (metros)
    else:
        gdf_cali_shp.set_crs('EPSG:4326', inplace=True)

# --- 7. CRUCE ESPACIAL POLIGONAL (SJOIN + DISSOLVE) ---
print("Realizando cruce espacial poligonal...")
COL_ID_MEDELLIN = gdf_medellin_shp.columns[0] 
COL_ID_CALI = gdf_cali_shp.columns[0]         

gdf_pred_med = gdf_pred_med.to_crs(gdf_medellin_shp.crs)
gdf_pred_cali = gdf_pred_cali.to_crs(gdf_cali_shp.crs)

gdf_final_med = gpd.sjoin(gdf_pred_med, gdf_medellin_shp[[COL_ID_MEDELLIN, 'geometry']], how='left', predicate='intersects')
gdf_final_cali = gpd.sjoin(gdf_pred_cali, gdf_cali_shp[[COL_ID_CALI, 'geometry']], how='left', predicate='intersects')

print(f"-> Puntos mapeados exitosamente en Medellín: {gdf_final_med[COL_ID_MEDELLIN].notna().sum():,}")
print(f"-> Puntos mapeados exitosamente en Cali: {gdf_final_cali[COL_ID_CALI].notna().sum():,}")

gdf_final_med = gdf_final_med[~gdf_final_med.index.duplicated(keep='first')]
gdf_final_cali = gdf_final_cali[~gdf_final_cali.index.duplicated(keep='first')]

gdf_mapa_medellin = gdf_final_med[[COL_ID_MEDELLIN, 'IVC_predicho', 'geometry']].dissolve(by=COL_ID_MEDELLIN, aggfunc={'IVC_predicho': 'mean'}).reset_index()
gdf_mapa_cali = gdf_final_cali[[COL_ID_CALI, 'IVC_predicho', 'geometry']].dissolve(by=COL_ID_CALI, aggfunc={'IVC_predicho': 'mean'}).reset_index()

gdf_mapa_medellin = gdf_mapa_medellin[gdf_mapa_medellin['IVC_predicho'].notna()].copy()
gdf_mapa_cali = gdf_mapa_cali[gdf_mapa_cali['IVC_predicho'].notna()].copy()

# --- 8. CARTOGRAFÍA ACADÉMICA Y RESPALDO EN S3 ---
def mapear_y_guardar(gdf_poligonos, nombre_ciudad, archivo_png, archivo_csv):
    if len(gdf_poligonos) == 0:
        print(f"❌ Error: El GeoDataFrame para {nombre_ciudad} está vacío. No se puede generar el mapa.")
        return

    print(f"Generando mapa cartográfico para {nombre_ciudad}...")
    gdf_plot = gdf_poligonos.to_crs(epsg=3857)
    
    minx, miny, maxx, maxy = gdf_plot.total_bounds
    buffer_x = (maxx - minx) * 0.12
    buffer_y = (maxy - miny) * 0.12
    minx, maxx = minx - buffer_x, maxx + buffer_x
    miny, maxy = miny - buffer_y, maxy + buffer_y

    aspecto_mapa = (maxy - miny) / (maxx - minx)
    ancho_figura = 12
    alto_figura = ancho_figura * aspecto_mapa

    fig, ax = plt.subplots(figsize=(ancho_figura, alto_figura), dpi=300)
    fondo_color = '#F8F9FA'
    fig.patch.set_facecolor(fondo_color)
    ax.set_facecolor(fondo_color)

    gdf_plot.plot(column='IVC_predicho', cmap='YlOrRd', linewidth=0, edgecolor='none', alpha=0.9, ax=ax, legend=False)
    ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron, alpha=0.8)

    ax.set_xlim(minx, maxx)
    ax.set_ylim(miny, maxy)
    ax.set_aspect('equal')
    ax.set_axis_off()
    plt.subplots_adjust(top=0.90, bottom=0.12, left=0.02, right=0.82)

    fig.text(0.02, 0.95, "Índice de Vulnerabilidad Climática (IVC) Predicho", fontsize=20, fontweight='bold', color='#1A1A1A')
    fig.text(0.02, 0.92, f"{nombre_ciudad} — Transferencia Espacial del Modelo", fontsize=14, color='#4D4D4D')

    vmin, vmax = gdf_plot['IVC_predicho'].min(), gdf_plot['IVC_predicho'].max()
    sm = plt.cm.ScalarMappable(cmap='YlOrRd', norm=plt.Normalize(vmin=vmin, vmax=vmax))
    sm._A = []
    cax = fig.add_axes([0.85, 0.30, 0.025, 0.40])
    cb = fig.colorbar(sm, cax=cax)
    cb.set_label("Nivel de Vulnerabilidad (IVC)", fontsize=12, labelpad=12, fontweight='bold', color='#1A1A1A')
    cb.outline.set_visible(False)
    cb.ax.tick_params(size=0, labelsize=11, colors='#1A1A1A')

    ax.annotate('N', xy=(0.95, 0.96), xytext=(0.95, 0.89), xycoords='axes fraction', arrowprops=dict(facecolor='#1A1A1A', edgecolor='none', width=2, headwidth=8, headlength=12), ha='center', va='top', fontsize=14, fontweight='bold', color='#1A1A1A')
    escala = ScaleBar(dx=1, units="m", location='lower right', box_alpha=0.7, box_color='white', color='#1A1A1A', font_properties={'size': 11, 'weight': 'bold'}, border_pad=1)
    ax.add_artist(escala)

    texto_metadatos = (
        "NOTAS METODOLÓGICAS:\n"
        f"• Algoritmo Predictivo: LightGBM (Transferencia de Dominio). Universo analizado: {gdf_plot.shape[0]:,} polígonos catastrales.\n"
        "• Tratamiento de datos: Valores atípicos excluidos mediante rango intercuartílico (IQR). Proyección: Web Mercator (EPSG:3857)."
    )
    fig.text(0.02, 0.02, texto_metadatos, fontsize=10, color='#4D4D4D', ha='left', linespacing=1.6)

    plt.savefig(archivo_png, dpi=300, bbox_inches='tight', pad_inches=0.1, facecolor=fondo_color)
    plt.show()

    gdf_poligonos.drop(columns=['geometry']).to_csv(archivo_csv, index=False)
    wr.s3.upload(archivo_png, f"s3://{bucket_salida}/mapas/{archivo_png}")
    wr.s3.upload(archivo_csv, f"s3://{bucket_salida}/datos/{archivo_csv}")
    print(f"✅ Resultados de {nombre_ciudad} respaldados exitosamente en S3.")

# Ejecución para Medellín y Cali
mapear_y_guardar(gdf_mapa_medellin, "Medellín, Antioquia", "mapa_ivc_medellin.png", "datos_ivc_medellin.csv")
mapear_y_guardar(gdf_mapa_cali, "Cali, Valle del Cauca", "mapa_ivc_cali.png", "datos_ivc_cali.csv")
```

```python
import os
import zipfile
import glob
import boto3
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import awswrangler as wr
from pycaret.regression import load_model, predict_model
import matplotlib.pyplot as plt
import contextily as ctx
from matplotlib_scalebar.scalebar import ScaleBar

# --- 1. CONFIGURACIÓN INICIAL ---
bucket_entrada = 'mapastiposhapemanzanascalimedellin'
bucket_salida = 'prediccionesbogota'

# --- 2. CARGA DEL MODELO Y DATOS TABULARES ---
print("Descargando modelo entrenado...")
wr.s3.download(path='s3://databogota/modelos/modelo_ivc_bogota_final_tuned.pkl', local_file='modelo_ivc_bogota_final_tuned.pkl')
modelo_produccion = load_model('modelo_ivc_bogota_final_tuned')

print("Cargando datos CSV de Medellín y Cali...")
df_medellin = wr.s3.read_csv(f's3://{bucket_entrada}/CLEAN DATA PARA MODELAR/MEDELLIN CALI/df_medellin_ivc_filtrado.csv')
df_cali = wr.s3.read_csv(f's3://{bucket_entrada}/CLEAN DATA PARA MODELAR/MEDELLIN CALI/df_cali_ivc_filtrado.csv')

# --- 3. ALINEACIÓN INTELIGENTE DE TIPOS DE DATOS ---
def alinear_con_modelo(df_nuevo, modelo):
    df_resultado = df_nuevo.copy()
    try:
        columnas_esperadas = modelo.steps[0][1].feature_names_in_
    except Exception:
        pipeline_obj = modelo.named_steps.get('actual_estimator', modelo)
        columnas_esperadas = getattr(pipeline_obj, 'feature_names_in_', df_nuevo.columns)

    for col in columnas_esperadas:
        if col not in df_resultado.columns:
            if col in ['NMB_LC_CM', 'TP_LC_CM']:
                df_resultado[col] = 'Desconocido'
            else:
                df_resultado[col] = 0.0 
                
    return df_resultado[[c for c in columnas_esperadas if c in df_resultado.columns]]

df_medellin_listo = alinear_con_modelo(df_medellin, modelo_produccion)
df_cali_listo = alinear_con_modelo(df_cali, modelo_produccion)

for df_orig, df_dest in zip([df_medellin, df_cali], [df_medellin_listo, df_cali_listo]):
    if 'LATITUD' in df_orig.columns:
        df_dest['LATITUD'] = df_orig['LATITUD']
    if 'LONGITUD' in df_orig.columns:
        df_dest['LONGITUD'] = df_orig['LONGITUD']

# --- 4. INFERENCIA DEL MODELO ---
print("Ejecutando predicciones para Medellín...")
pred_medellin = predict_model(modelo_produccion, data=df_medellin_listo)
pred_medellin.rename(columns={'prediction_label': 'IVC_predicho'}, inplace=True)

print("Ejecutando predicciones para Cali...")
pred_cali = predict_model(modelo_produccion, data=df_cali_listo)
pred_cali.rename(columns={'prediction_label': 'IVC_predicho'}, inplace=True)

# --- 5. UBICACIÓN ESPACIAL (PUNTOS) ---
print("Mapeando coordenadas espaciales...")
gdf_pred_med = gpd.GeoDataFrame(
    pred_medellin, 
    geometry=[Point(xy) for xy in zip(pred_medellin['LONGITUD'], pred_medellin['LATITUD'])], 
    crs='EPSG:4326'
)
gdf_pred_cali = gpd.GeoDataFrame(
    pred_cali, 
    geometry=[Point(xy) for xy in zip(pred_cali['LONGITUD'], pred_cali['LATITUD'])], 
    crs='EPSG:4326'
)

# --- 6. EXTRACCIÓN DE SHAPEFILES Y GEODATABASES ---
print("Procesando geometrías base de Medellín...")
wr.s3.download(f's3://{bucket_entrada}/shp_lote_del_predio.zip', local_file='medellin.zip')
if os.path.exists('shp_medellin'):
    import shutil
    shutil.rmtree('shp_medellin')
os.makedirs('shp_medellin', exist_ok=True)
with zipfile.ZipFile('medellin.zip', 'r') as z:
    z.extractall('shp_medellin')

rutas_shp_med = glob.glob('shp_medellin/**/*.shp', recursive=True)
if rutas_shp_med:
    gdf_medellin_shp = gpd.read_file(rutas_shp_med[0])
    print(f"✅ Shapefile de Medellín cargado correctamente.")
else:
    raise FileNotFoundError("❌ No se encontró ningún archivo .shp en Medellín.")

print("Procesando geodatabase base de Cali (.rar)...")
wr.s3.download(f's3://{bucket_entrada}/GEODATABASE CATASTRAL CON CORTE A ENERO 2024.rar', local_file='cali.rar')
if os.path.exists('shp_cali'):
    import shutil
    shutil.rmtree('shp_cali')
os.makedirs('shp_cali', exist_ok=True)

os.system('unrar x -y cali.rar shp_cali/')

rutas_gdb = glob.glob('shp_cali/**/*.gdb', recursive=True)
rutas_shp = glob.glob('shp_cali/**/*.shp', recursive=True)

if rutas_gdb:
    gdf_cali_shp = gpd.read_file(rutas_gdb[0], driver='OpenFileGDB')
    print(f"✅ Geodatabase de Cali cargada exitosamente: {rutas_gdb[0]}")
elif rutas_shp:
    gdf_cali_shp = gpd.read_file(rutas_shp[0])
    print(f"✅ Shapefile de Cali cargado exitosamente: {rutas_shp[0]}")
else:
    raise FileNotFoundError("❌ No se encontró ningún archivo .gdb o .shp tras descomprimir el .rar de Cali.")

if gdf_cali_shp.crs is None:
    if gdf_cali_shp.total_bounds[2] > 180:
        gdf_cali_shp.set_crs('EPSG:3115', inplace=True)
    else:
        gdf_cali_shp.set_crs('EPSG:4326', inplace=True)

# --- 7. CRUCE ESPACIAL DE MÁXIMA COBERTURA (BUFFER MÉTRICO PARA CALI) ---
print("Realizando cruce espacial optimizado con máxima cobertura...")
COL_ID_MEDELLIN = gdf_medellin_shp.columns[0] 
COL_ID_CALI = gdf_cali_shp.columns[0]         

# Medellín mantiene sjoin estándar por intersección
gdf_pred_med = gdf_pred_med.to_crs(gdf_medellin_shp.crs)
gdf_final_med = gpd.sjoin(gdf_pred_med, gdf_medellin_shp[[COL_ID_MEDELLIN, 'geometry']], how='left', predicate='intersects')

# CALI: Proyección a metros (EPSG:3115 - Magna-Sirgas Cali/Occidente)
CRS_METROS_CALI = 'EPSG:3115'
if gdf_cali_shp.crs.to_epsg() != 3115:
    gdf_cali_shp = gdf_cali_shp.to_crs(CRS_METROS_CALI)
gdf_pred_cali = gdf_pred_cali.to_crs(CRS_METROS_CALI)

# TRUCO MÁXIMO: Ampliamos las manzanas de Cali 18 metros hacia afuera 
# para capturar puntos en calles, andenes o con desfase de geocodificación.
print("Aplicando expansión geométrica (buffer de 18m) a las manzanas de Cali...")
gdf_cali_shp_buffered = gdf_cali_shp.copy()
gdf_cali_shp_buffered['geometry'] = gdf_cali_shp_buffered.geometry.buffer(18.0)

gdf_final_cali = gpd.sjoin(
    gdf_pred_cali, 
    gdf_cali_shp_buffered[[COL_ID_CALI, 'geometry']], 
    how='left', 
    predicate='intersects'
)

print(f"-> Puntos mapeados exitosamente en Medellín: {gdf_final_med[COL_ID_MEDELLIN].notna().sum():,}")
print(f"-> Puntos mapeados exitosamente en Cali (Con Buffer Máximo): {gdf_final_cali[COL_ID_CALI].notna().sum():,}")

gdf_final_med = gdf_final_med[~gdf_final_med.index.duplicated(keep='first')]
gdf_final_cali = gdf_final_cali[~gdf_final_cali.index.duplicated(keep='first')]

gdf_mapa_medellin = gdf_final_med[[COL_ID_MEDELLIN, 'IVC_predicho', 'geometry']].dissolve(by=COL_ID_MEDELLIN, aggfunc={'IVC_predicho': 'mean'}).reset_index()

# Para el disolver de Cali, usamos las geometrías originales (sin buffer) para mantener la cartografía limpia y exacta
gdf_final_cali_original = gdf_final_cali.merge(gdf_cali_shp[[COL_ID_CALI, 'geometry']], on=COL_ID_CALI, suffixes=('_buff', ''))
# Reemplazamos la geometría por la original
gdf_final_cali_original['geometry'] = gdf_final_cali_original['geometry'].fillna(gdf_final_cali_original['geometry_buff'])

gdf_mapa_cali = gdf_final_cali_original[[COL_ID_CALI, 'IVC_predicho', 'geometry']].dissolve(by=COL_ID_CALI, aggfunc={'IVC_predicho': 'mean'}).reset_index()

gdf_mapa_medellin = gdf_mapa_medellin[gdf_mapa_medellin['IVC_predicho'].notna()].copy()
gdf_mapa_cali = gdf_mapa_cali[gdf_mapa_cali['IVC_predicho'].notna()].copy()

# --- 8. CARTOGRAFÍA ACADÉMICA Y RESPALDO EN S3 ---
def mapear_y_guardar(gdf_poligonos, nombre_ciudad, archivo_png, archivo_csv):
    if len(gdf_poligonos) == 0:
        print(f"❌ Error: El GeoDataFrame para {nombre_ciudad} está vacío. No se puede generar el mapa.")
        return

    print(f"Generando mapa cartográfico para {nombre_ciudad}...")
    gdf_plot = gdf_poligonos.to_crs(epsg=3857)
    
    minx, miny, maxx, maxy = gdf_plot.total_bounds
    buffer_x = (maxx - minx) * 0.12
    buffer_y = (maxy - miny) * 0.12
    minx, maxx = minx - buffer_x, maxx + buffer_x
    miny, maxy = miny - buffer_y, maxy + buffer_y

    aspecto_mapa = (maxy - miny) / (maxx - minx)
    ancho_figura = 12
    alto_figura = ancho_figura * aspecto_mapa

    fig, ax = plt.subplots(figsize=(ancho_figura, alto_figura), dpi=300)
    fondo_color = '#F8F9FA'
    fig.patch.set_facecolor(fondo_color)
    ax.set_facecolor(fondo_color)

    gdf_plot.plot(column='IVC_predicho', cmap='YlOrRd', linewidth=0, edgecolor='none', alpha=0.9, ax=ax, legend=False)
    ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron, alpha=0.8)

    ax.set_xlim(minx, maxx)
    ax.set_ylim(miny, maxy)
    ax.set_aspect('equal')
    ax.set_axis_off()
    plt.subplots_adjust(top=0.90, bottom=0.12, left=0.02, right=0.82)

    fig.text(0.02, 0.95, "Índice de Vulnerabilidad Climática (IVC) Predicho", fontsize=20, fontweight='bold', color='#1A1A1A')
    fig.text(0.02, 0.92, f"{nombre_ciudad} — Transferencia Espacial del Modelo", fontsize=14, color='#4D4D4D')

    vmin, vmax = gdf_plot['IVC_predicho'].min(), gdf_plot['IVC_predicho'].max()
    sm = plt.cm.ScalarMappable(cmap='YlOrRd', norm=plt.Normalize(vmin=vmin, vmax=vmax))
    sm._A = []
    cax = fig.add_axes([0.85, 0.30, 0.025, 0.40])
    cb = fig.colorbar(sm, cax=cax)
    cb.set_label("Nivel de Vulnerabilidad (IVC)", fontsize=12, labelpad=12, fontweight='bold', color='#1A1A1A')
    cb.outline.set_visible(False)
    cb.ax.tick_params(size=0, labelsize=11, colors='#1A1A1A')

    ax.annotate('N', xy=(0.95, 0.96), xytext=(0.95, 0.89), xycoords='axes fraction', arrowprops=dict(facecolor='#1A1A1A', edgecolor='none', width=2, headwidth=8, headlength=12), ha='center', va='top', fontsize=14, fontweight='bold', color='#1A1A1A')
    escala = ScaleBar(dx=1, units="m", location='lower right', box_alpha=0.7, box_color='white', color='#1A1A1A', font_properties={'size': 11, 'weight': 'bold'}, border_pad=1)
    ax.add_artist(escala)

    texto_metadatos = (
        "NOTAS METODOLÓGICAS:\n"
        f"• Algoritmo Predictivo: LightGBM (Transferencia de Dominio). Universo analizado: {gdf_plot.shape[0]:,} polígonos catastrales.\n"
        "• Tratamiento de datos: Valores atípicos excluidos mediante rango intercuartílico (IQR). Proyección: Web Mercator (EPSG:3857)."
    )
    fig.text(0.02, 0.02, texto_metadatos, fontsize=10, color='#4D4D4D', ha='left', linespacing=1.6)

    plt.savefig(archivo_png, dpi=300, bbox_inches='tight', pad_inches=0.1, facecolor=fondo_color)
    plt.show()

    gdf_poligonos.drop(columns=['geometry']).to_csv(archivo_csv, index=False)
    wr.s3.upload(archivo_png, f"s3://{bucket_salida}/mapas/{archivo_png}")
    wr.s3.upload(archivo_csv, f"s3://{bucket_salida}/datos/{archivo_csv}")
    print(f"✅ Resultados de {nombre_ciudad} respaldados exitosamente en S3.")

# Ejecución para Medellín y Cali
mapear_y_guardar(gdf_mapa_medellin, "Medellín, Antioquia", "mapa_ivc_medellin.png", "datos_ivc_medellin.csv")
mapear_y_guardar(gdf_mapa_cali, "Cali, Valle del Cauca", "mapa_ivc_cali.png", "datos_ivc_cali.csv")
```
